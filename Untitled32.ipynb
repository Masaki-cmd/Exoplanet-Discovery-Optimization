## 1. データ読み込みと初期探索

まず、アップロードされた `cumulative.csv` ファイルをPandas DataFrameとして読み込み、データの構造と内容を把握します。

```python
import pandas as pd

# CSVファイルを読み込む
try:
    df = pd.read_csv('/content/cumulative.csv')
    print('データが正常に読み込まれました。')
except FileNotFoundError:
    print('エラー: cumulative.csvファイルが見つかりません。ファイルが正しくアップロードされているか確認してください。')
    df = None

if df is not None:
    # データの最初の5行を表示
    print('\nデータの最初の5行:')
    display(df.head())

    # データの基本情報を表示（列名、非Null数、データ型）
    print('\nデータの基本情報:')
    df.info()

    # 統計情報の要約を表示
    print('\nデータの統計情報の要約:')
    display(df.describe(include='all'))
```

## 2. データクリーニングと欠損値の確認

### 2.1. 欠損値の確認

まず、データセット内の各列にどの程度の欠損値（`NaN`）が存在するかを確認します。欠損値の割合が高い列は、そのままでは分析に使用できない場合や、データの解釈に影響を与える可能性があるため、適切な処理が必要です。

```python
# 各列の欠損値の数を計算
missing_values = df.isnull().sum()

# 各列の欠損値の割合を計算
missing_percentage = (df.isnull().sum() / len(df)) * 100

# 欠損値がある列をフィルタリングし、割合とともにDataFrameにまとめる
missing_info = pd.DataFrame({
    'Missing Count': missing_values,
    'Missing Percentage': missing_percentage
})

# 欠損値がある列のみを表示し、割合の降順でソート
missing_info = missing_info[missing_info['Missing Count'] > 0].sort_values(by='Missing Percentage', ascending=False)

print('\n欠損値のある列とその割合:')
display(missing_info)
```

### 2.2. 欠損値に対するクリーニング方針の提案

上記の欠損値の確認結果に基づき、以下のクリーニング方針を提案します。

**現状の課題認識:**

*   `koi_teq_err1` と `koi_teq_err2` は100%欠損しており、全く情報がないため、これらの列は削除するべきです。
*   `kepler_name` は76%以上の欠損があり、惑星に正式名称がまだ与えられていない、あるいは確認されていないケースが多く、この情報がKGI（収益最大化）に直接結びつく可能性は低いと判断できます。現時点では削除を検討します。
*   その他の列については、欠損率が比較的低いものが多く、特に `koi_score`, `koi_slogg_err2`, `koi_slogg_err1`, `koi_srad_err2`, `koi_srad_err1`, `koi_steff_err1`, `koi_steff_err2` などはエラーに関する情報であり、分析の目的に応じて扱いを検討する必要があります。これらは数値データであるため、平均値や中央値で補完することも可能ですが、エラー値であるため0で埋める選択肢もあります。
*   `koi_pdisposition` や `koi_disposition` はカテゴリカルデータであり、欠損値はありません。

**クリーニング方針:**

1.  **高欠損率の列の削除:**
    *   `koi_teq_err1`, `koi_teq_err2` (100%欠損): これらの列は分析に利用できないため、削除します。
    *   `kepler_name` (76%欠損): この列も欠損率が高く、直接的なビジネス分析への貢献が不薄いため削除します。

2.  **その他の数値データの欠損値補完（仮説設定が必要）:**
    *   残りの数値データ（例: `koi_score`, `koi_period_err1`, `koi_impact`, `koi_depth`, `koi_prad`, `koi_teq` など）については、ビジネス分析のKGIにどのように結びつけるかという具体的な仮説に基づいて、補完方法を決定します。
    *   **仮の補完方法:** 一旦、これらの数値データは、それぞれの列の**中央値**で補完することを提案します。中央値は外れ値の影響を受けにくく、データの分布を大きく歪める可能性が低いとされているためです。ただし、この補完方法がビジネス課題解決に最も適しているかどうかは、後続の分析で検証する必要があります。

この方針で進めることで、データセットの「汚さ」を処理し、整合性を保ちながら、次のステップであるデータ分析に進むことができます。特に、`koi_teq_err1`, `koi_teq_err2`, `kepler_name` の削除は、データの質を向上させる上で重要な一歩です。

```python
# クリーニング方針に従って列を削除
df_cleaned = df.drop(columns=['koi_teq_err1', 'koi_teq_err2', 'kepler_name'])

print('以下の列を削除しました: koi_teq_err1, koi_teq_err2, kepler_name')

# 残りの数値列の欠損値を中央値で補完
for column in df_cleaned.select_dtypes(include=['float64', 'int64']).columns:
    if df_cleaned[column].isnull().any():
        median_val = df_cleaned[column].median()
        df_cleaned[column].fillna(median_val, inplace=True)
        print(f'{column} の欠損値を中央値 {median_val} で補完しました。')

# クリーニング後のデータフレームの欠損値を再確認
print('\nクリーニング後の欠損値の確認:')
missing_values_after_cleaning = df_cleaned.isnull().sum()
missing_percentage_after_cleaning = (df_cleaned.isnull().sum() / len(df_cleaned)) * 100

missing_info_after_cleaning = pd.DataFrame({
    'Missing Count': missing_values_after_cleaning,
    'Missing Percentage': missing_percentage_after_cleaning
})

display(missing_info_after_cleaning[missing_info_after_cleaning['Missing Count'] > 0])

print('\nクリーニング後のデータフレームの基本情報:')
df_cleaned.info()
```

```python
# koi_tce_delivname の欠損値を 'Unknown' で補完
df_cleaned['koi_tce_delivname'].fillna('Unknown', inplace=True)

print("koi_tce_delivname の欠損値を 'Unknown' で補完しました。")

# 最終的な欠損値の確認
print('\n最終クリーニング後の欠損値の確認:')
missing_info_final = pd.DataFrame({
    'Missing Count': df_cleaned.isnull().sum(),
    'Missing Percentage': (df_cleaned.isnull().sum() / len(df_cleaned)) * 100
})
display(missing_info_final[missing_info_final['Missing Count'] > 0])

print('\nすべての欠損値が処理されました。データクリーニングが完了しました。')
```

## 3. 課題定義と仮説に基づいた分析

### 3.2. `koi_disposition` と `koi_pdisposition` の数値化

`koi_disposition` および `koi_pdisposition` はカテゴリカルデータであるため、機械学習モデルなどで扱えるように数値化します。ここでは、各カテゴリをバイナリ（0または1）の新しい列に変換するOne-Hot Encodingを適用します。

**この処理を行う意図:**

*   カテゴリカルなターゲット変数を数値化することで、相関分析やモデル構築が可能になります。
*   One-Hot Encodingは、各カテゴリ間に順序関係がない場合に適しており、モデルがカテゴリを平等に扱うことができます。

**結果から導かれる提案の論理:**

この数値化により、後続の分析で「CONFIRMED」惑星がどのような特徴を持つのか、また「FALSE POSITIVE」がなぜそのように分類されるのかを、具体的な数値データとして掘り下げていく基盤を構築します。例えば、「CONFIRMED」に変換された新しい列と他の数値特徴量との相関を分析することで、どの天文学的特徴が確認済み惑星であることと強く関連しているかを特定し、効率的なスクリーニングプロセスに繋げることができます。

```python
# One-Hot Encoding を適用してカテゴリカルデータを数値化
# 'koi_disposition' を数値化
df_encoded = pd.get_dummies(df_cleaned, columns=['koi_disposition'], prefix='disposition')

# 'koi_pdisposition' も数値化（必要であれば、こちらもOne-Hot Encoding）
df_encoded = pd.get_dummies(df_encoded, columns=['koi_pdisposition'], prefix='pdisposition')

print('カテゴリカル変数を数値化しました。')
print('\n数値化後のデータフレームの最初の5行:')
display(df_encoded.head())

print('\n数値化後のデータフレームの列情報:')
df_encoded.info()
```

### 3.3. 特徴量の相関分析

`koi_disposition`が数値化されたので、次に他の数値特徴量との相関関係を分析します。これにより、「CONFIRMED」な惑星であることと、特に強く関連する天文学的特徴を特定します。

**この分析を行う意図:**

*   「CONFIRMED」惑星の検出に貢献する可能性のある主要な特徴量を特定します。
*   相関の高い特徴量は、将来的に「CONFIRMED」惑星のスクリーニング精度を向上させるための指標となり、分析リソースを効率的に配分するために役立ちます。

**結果から導かれる提案の論理:**

もし特定の物理量（例：`koi_prad` (惑星半径)、`koi_period` (軌道周期)）が「CONFIRMED」と高い正の相関を持つ場合、それらの特徴を持つ候補に優先的にリソースを割り当てることで、確認作業の成功率を高め、結果として収益を最大化できる可能性があります。逆に、「FALSE POSITIVE」と高い相関を持つ特徴量があれば、そのような候補は初期段階で除外することで、無駄なコストを削減できるでしょう。

```python
# 'disposition_CONFIRMED' と他の数値特徴量との相関を計算

# まず、One-Hot Encodingで生成されたbool型の列をint型に変換し、数値型の列として含める
df_corr_ready = df_encoded.copy()
for col in df_corr_ready.select_dtypes(include='bool').columns:
    df_corr_ready[col] = df_corr_ready[col].astype(int)

# すべての数値型の列（float64, int64, および変換後のbool型）を選択
numeric_cols_for_corr = df_corr_ready.select_dtypes(include=['float64', 'int64']).columns

# 'disposition_CONFIRMED' との相関を計算
correlation_with_confirmed = df_corr_ready[numeric_cols_for_corr].corr()['disposition_CONFIRMED'].sort_values(ascending=False)

print('\n"disposition_CONFIRMED" と他の数値特徴量の相関:')
print(correlation_with_confirmed)

# 相関の高い特徴量を可視化
plt.figure(figsize=(12, 8))
sns.barplot(x=correlation_with_confirmed.index, y=correlation_with_confirmed.values, palette='viridis')
plt.title('Correlation with disposition_CONFIRMED')
plt.xlabel('Features')
plt.ylabel('Correlation Coefficient')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()

print("\n### このグラフを作成した意図:")
print("この棒グラフは、「CONFIRMED」な惑星であること（disposition_CONFIRMEDが1であること）と、データセット内の各数値特徴量との間のピアソン相関係数を示しています。どの天文学的特徴が「CONFIRMED」という分類と強く関連しているかを一覧で把握するために作成しました。これにより、特に注目すべき特徴量を特定し、後続のより詳細な分析の方向性を定めることができます。")
print("\n### 結果から導かれる提案の論理:")
print("相関係数の絶対値が高い特徴量は、disposition_CONFIRMEDとの関連性が強いことを意味します。例えば、正の相関が強い特徴は「CONFIRMED」惑星に多く見られる特性であり、負の相関が強い特徴は「FALSE POSITIVE」に多く見られる特性である可能性があります。この情報に基づき、高相関の特徴量を持つ候補に分析リソースを集中させたり、低相関あるいは逆相関の特徴量を持つ候補を早期にスクリーニングから外したりすることで、確認プロセスの効率を大幅に向上させることが提案できます。これにより、より多くの「CONFIRMED」惑星を効率的に特定し、収益機会を最大化し、無駄なコストを削減することに繋がります。")
```

### 3.4. ランダムフォレストによる特徴量の重要度分析

相関分析に加えて、機械学習モデルの一つであるランダムフォレストを用いて、`CONFIRMED`惑星の分類に対する各特徴量の重要度を評価します。ランダムフォレストは、多数の決定木を構築し、それらの結果を統合することで、予測性能が高く、過学習に強い特徴があります。また、特徴量間の非線形な関係も捉えることができます。

**この分析を行う意図:**

*   `disposition_CONFIRMED`を予測する上で、どの特徴量が最も影響力を持つかを特定します。
*   相関分析だけでは捉えきれない、より複雑な特徴量間の関係性を考慮した重要度を把握します。

**結果から導かれる提案の論理:**

ランダムフォレストによって高い重要度を持つと評価された特徴量は、`CONFIRMED`惑星の候補を効率的に特定するための強力な指標となります。これらの特徴量に焦点を当てることで、惑星候補のスクリーニングプロセスをさらに最適化し、資源の無駄を削減し、収益機会を最大化することが期待されます。

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
import numpy as np

# 'df_corr_ready' を使用して特徴量とターゲットを定義
# ターゲット変数 'disposition_CONFIRMED' を抽出
y = df_corr_ready['disposition_CONFIRMED']

# ターゲットと関連する（または一意の識別子である）列を除外して特徴量Xを作成
# 'kepoi_name' はオブジェクト型なので除外
# 'disposition_CANDIDATE', 'disposition_FALSE POSITIVE', 'pdisposition_CANDIDATE', 'pdisposition_FALSE POSITIVE'
# は 'disposition_CONFIRMED' と直接関連するため、特徴量から除外
# 'rowid' と 'kepid' も識別子なので除外
excluded_cols = ['disposition_CONFIRMED', 'disposition_CANDIDATE', 'disposition_FALSE POSITIVE',
                 'pdisposition_CANDIDATE', 'pdisposition_FALSE POSITIVE', 'kepoi_name', 'rowid', 'kepid']
X = df_corr_ready.drop(columns=excluded_cols)

# Xに残っているオブジェクト型の列を確認（'koi_tce_delivname' が残っているはず）
object_cols_in_X = X.select_dtypes(include='object').columns
if len(object_cols_in_X) > 0:
    # 'koi_tce_delivname' を再度One-Hot Encodingで数値化
    X = pd.get_dummies(X, columns=object_cols_in_X, prefix='tce_delivname')

print('特徴量とターゲット変数を準備しました。')
print(f'特徴量の数: {X.shape[1]}')
print(f'ターゲット変数のサンプル数: {y.shape[0]}')

# データを訓練セットとテストセットに分割
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

print(f'訓練セットのサンプル数: {X_train.shape[0]}')
print(f'テストセットのサンプル数: {X_test.shape[0]}')
```

```python
# ランダムフォレストモデルの初期化と訓練
# クラスの不均衡があるため、class_weight='balanced' を使用
model = RandomForestClassifier(n_estimators=100, random_state=42, class_weight='balanced')
model.fit(X_train, y_train)

print('ランダムフォレストモデルの訓練が完了しました。')

# テストセットでの予測と精度評価
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f'モデルのテスト精度: {accuracy:.4f}')
```

```python
# 特徴量の重要度を抽出
feature_importances = pd.Series(model.feature_importances_, index=X.columns).sort_values(ascending=False)

print('\n特徴量の重要度 (Random Forest):')
print(feature_importances.head(20)) # 上位20個の特徴量を表示

# 特徴量の重要度を可視化 (上位20個)
plt.figure(figsize=(12, 8))
sns.barplot(x=feature_importances.head(20).index, y=feature_importances.head(20).values, palette='viridis')
plt.title('Feature Importances for disposition_CONFIRMED (Random Forest)')
plt.xlabel('Features')
plt.ylabel('Importance Score')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()

print('\n### このグラフを作成した意図:')
print('この棒グラフは、ランダムフォレストモデルが「CONFIRMED」惑星を分類する際に、各特徴量をどの程度重要と判断したかを示しています。相関分析だけでは捉えきれない、特徴量間の複雑な相互作用も考慮した上での重要度を視覚化することで、真に予測に貢献する特徴量を特定する目的で作成しました。')
print('\n### 結果から導かれる提案の論理:')
print('ランダムフォレストの重要度上位に位置する特徴量は、`CONFIRMED`惑星を識別するための最も強力な情報源であると考えられます。これらの特徴量に特に注意を払い、データ収集、観測、および後続の分析リソースを集中させることで、検出プロセスの効率性を最大化し、収益を向上させることが可能です。また、重要度の低い特徴量は、モデルの複雑性を軽減したり、データ収集のコストを削減したりするために、将来的な分析から除外することも検討できます。これにより、より洗練された予測モデルを構築し、ビジネス上の意思決定を強化することができます。')
```

### 3.5. 重要度上位の特徴量の散布図分析

ランダムフォレストモデルで高い重要度を示した特徴量 (`koi_score`, `koi_model_snr`, `koi_steff_err1`) について、それらの間の関係性と `disposition_CONFIRMED` との関連性を散布図で可視化します。これにより、これらの特徴量がどのように「CONFIRMED」惑星と他の分類を区別しているかの傾向を視覚的に把握します。

**この分析を行う意図:**

*   最も重要な特徴量間の相互関係を理解し、分類境界がどのように形成されているかを探ります。
*   特定の領域に「CONFIRMED」惑星が集中しているか、あるいは「FALSE POSITIVE」が集中しているかを確認します。

**結果から導かれる提案の論理:**

### 3.6. ランダムフォレストのハイパーパラメータ調整による精度向上

モデルの予測性能をさらに高めるため、ランダムフォレストのハイパーパラメータ調整を行います。`GridSearchCV`を用いて、最適なパラメータの組み合わせを探索します。これにより、過学習を抑制しつつ、より高い汎化性能を持つモデルを構築できる可能性があります。

**この分析を行う意図:**

*   ランダムフォレストモデルの予測精度を体系的に向上させます。
*   最適なモデル設定を見つけ、ビジネス課題に対する解決策の信頼性を高めます。

**結果から導かれる提案の論理:**

ハイパーパラメータ調整によって改善されたモデルは、新規の惑星候補を「CONFIRMED」または「FALSE POSITIVE」により正確に分類できるようになります。これにより、以下のビジネス上の効果が期待されます。

*   **収益最大化:** 「CONFIRMED」と判断された候補が実際に確認済みとなる確率が高まり、発見された惑星を通じたサービス提供の機会が増加します。
*   **コスト最適化:** 「FALSE POSITIVE」と判断された候補への無駄な観測・分析リソースの投入が減少し、効率的な資源配分が可能になります。

```python
from sklearn.model_selection import GridSearchCV, StratifiedKFold

# ランダムフォレストモデルの初期化
rf_model = RandomForestClassifier(random_state=42, class_weight='balanced')

# 調整するハイパーパラメータの範囲を定義
param_grid = {
    'n_estimators': [100, 200, 300], # 決定木の数
    'max_depth': [10, 20, None],      # 決定木の最大深度 (Noneは制限なし)
    'min_samples_split': [2, 5, 10],  # ノードを分割するために必要な最小サンプル数
    'min_samples_leaf': [1, 2, 4]     # 葉ノードに必要な最小サンプル数
}

# StratifiedKFold を使用してクロスバリデーションを設定 (クラスの不均衡に対応)
skf = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)

# GridSearchCV の設定
grid_search = GridSearchCV(
    estimator=rf_model,
    param_grid=param_grid,
    cv=skf,
    scoring='accuracy', # 評価指標
    n_jobs=-1,          # 全てのCPUコアを使用
    verbose=1           # 進行状況を表示
)

print('GridSearchCV を開始します。この処理には時間がかかる場合があります...')

# GridSearchCV の実行
grid_search.fit(X_train, y_train)

print('\nGridSearchCV が完了しました。')

# 最適なハイパーパラメータとスコアを表示
print(f"最適なハイパーパラメータ: {grid_search.best_params_}")
print(f"最適なクロスバリデーション精度: {grid_search.best_score_:.4f}")

# 最適なモデルを取得
best_rf_model = grid_search.best_estimator_

# テストセットでの最終評価
y_pred_tuned = best_rf_model.predict(X_test)
accuracy_tuned = accuracy_score(y_test, y_pred_tuned)
print(f"調整後モデルのテスト精度: {accuracy_tuned:.4f}")

print('\n### このコードを作成した意図:')
print('このコードは、ランダムフォレストモデルのハイパーパラメータを最適化し、予測精度を向上させるために作成されました。GridSearchCVを使用することで、複数のハイパーパラメータの組み合わせを網羅的に探索し、交差検定を通じて最も性能の良いモデル設定を特定します。これにより、単一のデフォルト設定に依存するのではなく、データに最も適したモデルを構築できます。')
print('\n### 結果から導かれる提案の論理:')
print('調整されたモデルは、調整前のモデルよりも高い、または同等のテスト精度を示すはずです。この改善された精度により、新しい惑星候補を「CONFIRMED」または「FALSE POSITIVE」として分類する際の信頼性が向上します。この高精度なモデルを活用することで、惑星候補のスクリーニングプロセスをより効率的に行い、発見される「CONFIRMED」惑星の数を増やし、同時に無駄な観測リソースの投入を最小限に抑えることで、ビジネスの収益最大化とコスト最適化に貢献します。具体的には、このモデルの予測結果を、初期スクリーニングの最終判断層として導入することが推奨されます。')
```

**この分析を行う意図:**

### 3.8. 適合率と再現率のトレードオフ分析と最適な閾値の探索

予測閾値を調整することで、適合率（Precision）と再現率（Recall）がどのように変化するかを視覚的に理解することは、ビジネス目標に合致した最適なモデル運用戦略を策定する上で不可欠です。

**この分析を行う意図:**

*   異なる閾値におけるモデルの性能（適合率、再現率、F1スコア）の変化を定量的に把握します。
*   「CONFIRMED」惑星の発見機会の最大化（高再現率）と、偽陽性へのリソース投入削減（高適合率）というビジネス上の二つの目標の間で、最適なバランス点を見つけるための根拠を提供します。

**結果から導かれる提案の論理:**

この分析により、特定の閾値において適合率と再現率がどのようにトレードオフの関係にあるかが明確になります。例えば、適合率を重視して偽陽性を徹底的に排除したい場合は、適合率が高く保たれる閾値を選択します。一方、多少の偽陽性は許容しても、できるだけ多くの「CONFIRMED」惑星を見つけたい（収益機会を最大化したい）場合は、再現率が高く保たれる閾値を選択します。

F1スコアは適合率と再現率のバランスを示す指標であり、両方をある程度重視する場合の参考になります。このグラフを参考に、ビジネスにおけるリスク許容度とリターン（収益）のバランスを考慮し、最も適切な閾値を決定することが可能になります。これにより、より戦略的な意思決定に基づいた効率的な惑星候補のスクリーニングプロセスが確立され、結果として収益の最大化とコストの最適化が実現されます。

```python
from sklearn.metrics import precision_recall_curve, f1_score

# CONFIRMEDクラスの予測確率
conf_probabilities = best_rf_model.predict_proba(X_test)[:, 1]

# 適合率、再現率、閾値の計算
precisions, recalls, thresholds = precision_recall_curve(y_test, conf_probabilities)

# F1スコアの計算 (各閾値に対応する予測結果でF1スコアを計算するため、閾値と一致する形で予測を行う)
f1_scores = []
for t in thresholds:
    y_pred_at_threshold = (conf_probabilities >= t).astype(int)
    f1_scores.append(f1_score(y_test, y_pred_at_threshold, zero_division=0))

# F1スコアが最大となる閾値を特定
optimal_f1_idx = np.argmax(f1_scores)
optimal_threshold = thresholds[optimal_f1_idx]
optimal_precision = precisions[optimal_f1_idx]
optimal_recall = recalls[optimal_f1_idx]

print(f"最大F1スコア時の閾値: {optimal_threshold:.4f}")
print(f"最大F1スコア時の適合率: {optimal_precision:.4f}")
print(f"最大F1スコア時の再現率: {optimal_recall:.4f}")

# 適合率、再現率、F1スコアを閾値に対してプロット
plt.figure(figsize=(12, 7))
plt.plot(thresholds, precisions[:-1], label='Precision', linestyle='-', color='blue')
plt.plot(thresholds, recalls[:-1], label='Recall', linestyle='-', color='red')
plt.plot(thresholds, f1_scores, label='F1-score', linestyle='--', color='green')

plt.axvline(x=optimal_threshold, color='purple', linestyle=':', label=f'Optimal F1 Threshold ({optimal_threshold:.2f})')
plt.text(optimal_threshold + 0.02, 0.5, f'P:{optimal_precision:.2f}, R:{optimal_recall:.2f}', rotation=90, va='center', color='purple')

plt.title('Precision, Recall, and F1-score vs. Threshold')
plt.xlabel('Threshold')
plt.ylabel('Score')
plt.legend()
plt.grid(True)
plt.show()

print('''\n### このグラフを作成した意図:
このグラフは、モデルの分類閾値を0から1まで変化させたときに、適合率（Precision）、再現率（Recall）、そしてF1スコアがどのように変化するかを示しています。特に、ビジネス目標（収益最大化とコスト最適化）に応じて、どの閾値が最適であるかを視覚的に判断するためのツールとして作成されました。''')

print('''\n### 結果から導かれる提案の論理:
グラフから、閾値を上げると適合率が向上し（偽陽性が減少）、再現率が低下する（真陽性を見逃す可能性が増加）というトレードオフの関係が明確に確認できます。ビジネス目標が「偽陽性の徹底的な排除（コスト最適化）」であれば、適合率が急上昇する閾値や、F1スコアが最大になる閾値を参考にします。逆に、「CONFIRMED」惑星の発見数を最大化したい（収益最大化）のであれば、再現率が高く保たれる閾値を選択します。
F1スコアは適合率と再現率のバランスを示すため、両方をバランスよく重視したい場合の参考になります。この視覚的な情報を用いて、企業の戦略やリスク許容度に基づいて最適な閾値を設定し、予測モデルの運用方針を決定することで、効率的なリソース配分とビジネス成果の最大化を図ることが可能になります。''')
```

### 4.1. PrecisionとRecallを考慮したビジネス上の損益分岐点（収益最大化点）の算出

モデルの予測閾値を調整することで、適合率（Precision）と再現率（Recall）が変動し、これはビジネス上の収益とコストに直接的な影響を与えます。

*   **真陽性 (True Positive: TP):** 正しく「CONFIRMED」と予測され、実際に「CONFIRMED」であった場合。これは収益（例：研究サービス提供による収益）に貢献します。
*   **偽陽性 (False Positive: FP):** 「CONFIRMED」と誤って予測され、実際には「FALSE POSITIVE」であった場合。これはコスト（例：無駄な詳細観測、追加分析の人件費）となります。
*   **偽陰性 (False Negative: FN):** 「CONFIRMED」でないと誤って予測され、実際には「CONFIRMED」であった場合。これは機会損失（例：発見できたはずの惑星を見逃すことによる収益機会の損失）となります。
*   **真陰性 (True Negative: TN):** 正しく「FALSE POSITIVE」と予測され、実際に「FALSE POSITIVE」であった場合。これはコスト削減（例：無駄な観測を回避することによるコスト最適化）に貢献します。

これらの要素を考慮し、異なる閾値におけるビジネス上の純利益を計算することで、収益が最大化される「損益分岐点」または「最適運用点」を特定します。

**この分析を行う意図:**

*   モデルの技術的な性能指標（Precision, Recall）を、具体的なビジネスの収益とコストの観点から評価します。
*   企業の戦略的目標（収益最大化か、コスト最適化か）に基づいて、最も経済的に有利な予測閾値を特定するための定量的な根拠を提供します。

**結果から導かれる提案の論理:**

純利益を最大化する閾値を特定することで、企業はモデルの運用戦略を最適化できます。例えば、

*   **TP1件あたりの収益:** `R_TP`
*   **FP1件あたりのコスト:** `C_FP`
*   **FN1件あたりの機会損失:** `C_FN`
*   **TN1件あたりのコスト削減:** `S_TN`

と仮定した場合、総利益は `(TP * R_TP) - (FP * C_FP) - (FN * C_FN) + (TN * S_TN)` で計算されます。この計算を様々な閾値で行い、最も高い利益が得られる閾値が、ビジネス上の最適な運用閾値となります。この分析結果は、経営層がリスク許容度とリターンを考慮し、データドリブンな意思決定を行う上で不可欠な情報となります。

```python
from sklearn.metrics import confusion_matrix

# --- ビジネス上の仮定値（ユーザーが調整するべき項目） ---
# 1件のCONFIRMED惑星発見による収益（例: 研究サービス契約による平均収益）
REVENUE_PER_TRUE_POSITIVE = 1000  # 単位: ドル

# 1件のFALSE POSITIVEに対するコスト（例: 無駄な観測、分析にかかる人件費、機器使用料）
COST_PER_FALSE_POSITIVE = 200    # 単位: ドル

# 1件のCONFIRMED惑星を見逃すことによる機会損失（例: 潜在的な収益の逸失）
OPPORTUNITY_COST_PER_FALSE_NEGATIVE = 500 # 単位: ドル

# 1件のTRUE NEGATIVEによるコスト削減（例: FALSE POSITIVEを避けることによる直接的なコスト節約）
SAVINGS_PER_TRUE_NEGATIVE = 50     # 単位: ドル
# -----------------------------------------------------

# CONFIRMEDクラスの予測確率
conf_probabilities = best_rf_model.predict_proba(X_test)[:, 1]

# 各閾値における利益を計算
profits = []
thresholds_for_profit = np.linspace(0, 1, 100) # 0から1まで100段階で閾値を設定

for threshold in thresholds_for_profit:
    y_pred_at_threshold = (conf_probabilities >= threshold).astype(int)

    # 混同行列を計算 (y_testは真の値、y_pred_at_thresholdは予測値)
    tn, fp, fn, tp = confusion_matrix(y_test, y_pred_at_threshold).ravel()

    # 総利益を計算
    current_profit = (tp * REVENUE_PER_TRUE_POSITIVE) \
                     - (fp * COST_PER_FALSE_POSITIVE) \
                     - (fn * OPPORTUNITY_COST_PER_FALSE_NEGATIVE) \
                     + (tn * SAVINGS_PER_TRUE_NEGATIVE)
    profits.append(current_profit)

# 最大利益と、その時の閾値を特定
max_profit = max(profits)
optimal_profit_threshold = thresholds_for_profit[np.argmax(profits)]

print(f"最大利益: ${max_profit:,.2f}")
print(f"最大利益時の最適閾値: {optimal_profit_threshold:.4f}")

# 利益を閾値に対してプロット
plt.figure(figsize=(12, 7))
plt.plot(thresholds_for_profit, profits, marker='.', linestyle='-')
plt.axvline(x=optimal_profit_threshold, color='red', linestyle='--', label=f'Optimal Profit Threshold ({optimal_profit_threshold:.2f})')
plt.title('Profit vs. Prediction Threshold')
plt.xlabel('Prediction Threshold')
plt.ylabel('Total Profit ($)')
plt.grid(True)
plt.legend()
plt.show()

print('\n### このグラフを作成した意図:')
print('このグラフは、モデルの予測閾値を変更した際に、ビジネス上の総利益がどのように変化するかを視覚的に示すために作成されました。技術的な指標だけでなく、具体的な収益とコストのバランスを考慮に入れることで、ビジネス目標に最も合致する最適な運用閾値を特定することを目的としています。')

print('\n### 結果から導かれる提案の論理:')
print('グラフのピークが示す閾値が、このモデルと現在のビジネス仮定値において最も利益を最大化する点です。この閾値を運用基準として採用することで、偽陽性への無駄なリソース投入を最小限に抑えつつ、真のCONFIRMED惑星からの収益を最大化することができます。ただし、REVENUE_PER_TRUE_POSITIVE, COST_PER_FALSE_POSITIVE, OPPORTUNITY_COST_PER_FALSE_NEGATIVE, SAVINGS_PER_TRUE_NEGATIVE の各値をビジネスの現実的な状況に合わせて正確に設定することが極めて重要です。これらの値が変更されれば、最適な閾値も変動するため、ビジネスモデルの変化に合わせて定期的な再評価が必要です。')
```
